```python
import numpy as np
from scipy.stats import chi2_contingency


# Data dari tabel
data = np.array([[8, 22, 15, 5],
                 [10, 28, 20, 7],
                 [12, 30, 20, 8]])


# Menghitung chi-kuadrat
chi2, p, dof, expected = chi2_contingency(data)


# Menampilkan hasil
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)


# Menentukan keputusan
alpha = 0.01
if p < alpha:
    print("Tolak H0: Terdapat hubungan antara kecukupan tidur dan kekuatan gowes.")
else:
    print("Gagal Tolak H0: Tidak terdapat hubungan antara kecukupan tidur dan kekuatan gowes.")
```

```python
import numpy as np
from scipy import stats


# Data sampel nilai dari 15 mahasiswa
data = [904, 920, 973, 1001, 1002, 1002, 1012, 1016, 1039, 1086, 1140, 1146, 1168, 1233, 1255, 1348]


# Melakukan Uji Kolmogorov-Smirnov
stat, p_value = stats.kstest(data, 'norm', args=(np.mean(data), np.std(data, ddof=1)))


print("Statistik Uji Kolmogorov-Smirnov:", stat)
print("p-value:", p_value)


# Menentukan apakah data berdistribusi normal atau tidak
alpha = 0.05
if p_value > alpha:
    print("Data berdistribusi normal (gagal menolak H0)")
else:
    print("Data tidak berdistribusi normal (menolak H0)")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

# Data sampel nilai 15 mahasiswa
data = [12, 25, 45, 67, 43, 33, 24, 45, 34, 11, 8, 34, 67, 99, 22]
data_sorted = sorted(data)
n = len(data_sorted)

# Hitung statistik deskriptif
mean = np.mean(data_sorted)
std_dev = np.std(data_sorted, ddof=1)

# Buat tabel perhitungan
results = []
for i, xi in enumerate(data_sorted):
    z = (xi - mean) / std_dev
    ft_xi = norm.cdf(xi, mean, std_dev)
    fs_xi = (i + 1) / n
    diff = abs(ft_xi - fs_xi)

    results.append({
        'Xi': xi,
        'z': round(z, 4),
        'Ft(xi)': round(ft_xi, 4),
        'Fs(xi)': round(fs_xi, 4),
        '|Ft(xi)-Fs(xi)|': round(diff, 4)
    })

# Tampilkan tabel
df = pd.DataFrame(results)
print(df)

# Hitung D maksimum
D_max = max([result['|Ft(xi)-Fs(xi)|'] for result in results])
D_tabel = 0.338

print(f"\nD hitung: {D_max:.4f}")
print(f"D tabel: {D_tabel}")

# Keputusan
if D_max > D_tabel:
    print("TOLAK H0 - Data tidak berdistribusi normal")
else:
    print("GAGAL MENOLAK H0 - Data berdistribusi normal")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, chi2

# Data tabel kontingensi
data = np.array([
    [14, 6, 9],   # Baik
    [10, 16, 10], # Cukup
    [2, 13, 20]   # Jelek
])

# Buat DataFrame
df = pd.DataFrame(data,
                  index=['Baik', 'Cukup', 'Jelek'],
                  columns=['Tinggi', 'Sedang', 'Rendah'])
print("TABEL KONTINGENSI:")
print(df)

# Hitung Chi-Square
chi2_stat, p_value, dof, expected = chi2_contingency(data)

# Nilai kritis
alpha = 0.05
chi2_tabel = chi2.ppf(1 - alpha, dof)

print(f"\nChi-Square hitung: {chi2_stat:.4f}")
print(f"Chi-Square tabel: {chi2_tabel:.4f}")
print(f"p-value: {p_value:.6f}")

# Keputusan
if chi2_stat > chi2_tabel:
    print("TOLAK H0 - Ada hubungan signifikan antara pendapatan dan kualitas bahan makanan")
else:
    print("GAGAL MENOLAK H0 - Tidak ada hubungan signifikan antara pendapatan dan kualitas bahan makanan")
```
