```python
import pandas as pd

# Membaca data dari clipboard
rafi2 = pd.read_clipboard()

# Menampilkan data
print(rafi2)
```

```python
# Memfilter data berdasarkan 'Bedrooms' == 2
filtered_rafi2 = rafi2[rafi2['Tinggi Badan'] == 171]

# Menampilkan hasil setelah filter
print("\nData setelah filter (Tinggi Badan == 171):")
print(filtered_rafi2)
```

```python
# Mengubah kolom Tinggi_Badan
rafi2['Tinggi Badan'] = rafi2['Tinggi Badan'].apply(lambda x: 'Tinggi' if x > 160 else 'Pendek')

print(rafi2)
```

```python
# Menambahkan kolom "Jurusan" berdasarkan nilai "Angkatan"
rafi2['Jurusan'] = rafi2['Angkatan'].apply(lambda x: 'Infor2020' if x == 2020 else 'Infor2021')

# Menambahkan kolom "Fakultas" dengan nilai "FTI"
rafi2['Fakultas'] = 'FTI'

# Menampilkan DataFrame setelah penambahan kolom baru
print("\nData Setelah Penambahan Kolom:")

print(rafi2.to_string(index=False))  
```

```python
# Menghapus kolom 'newvariable' dari DataFrame
rafi2 = rafi2.drop(columns=['Fakultas'])

# Menampilkan DataFrame 'rafi' setelah kolom dihapus
print(rafi2.to_string(index=False))  
```

```python
# Gabung kolom Nama dan Gender (kolom 1 dan 2)
rafi2['Nama_Gender'] = rafi2.iloc[:, 0] + ' ' + rafi2.iloc[:, 1]

# Gabung kolom Angkatan dan Tinggi Badan (kolom 3 dan 4)
rafi2['Angkatan_Tinggi'] = rafi2.iloc[:, 2].astype(str) + ' ' + rafi2.iloc[:, 3].astype(str)

# Gabungkan kedua kolom gabungan tersebut
rafi2['Gabungan'] = rafi2['Nama_Gender'] + ' | ' + rafi2['Angkatan_Tinggi']

# Menampilkan DataFrame setelah penggabungan kolom
print(rafi2[['Nama_Gender', 'Angkatan_Tinggi', 'Gabungan']])
```

```python
# Mengambil baris 1-5 (indeks 0 sampai 4) dan baris 25-30 (indeks 24 sampai 29)
gabungan_baris_1_5 = rafi2.iloc[0:5]
gabungan_baris_25_30 = rafi2.iloc[24:30]

# Menggabungkan dua DataFrame tersebut
gabungan_baris = pd.concat([gabungan_baris_1_5, gabungan_baris_25_30], ignore_index=True)

# Menampilkan hasil penggabungan
print(gabungan_baris)
```

```python
# Mengurutkan DataFrame berdasarkan kolom 'Waktu Perjalanan'
data_rafi2_sort = rafi2.sort_values(by='Waktu Perjalanan')

# Menampilkan DataFrame yang telah diurutkan
print(data_rafi2_sort)
```
