```python
import pandas as pd

# Membaca data dari clipboard
rafi = pd.read_clipboard(sep=',')

# Menampilkan data
print(rafi)
```

```python
# Memfilter data berdasarkan 'Bedrooms' == 2
filtered_rafi = rafi[rafi['Bedrooms'] == 2]

# Menampilkan hasil setelah filter
print("\nData setelah filter (Bedrooms == 2):")
print(filtered_rafi)
```

```python
rafi['Bathrooms'] = pd.to_numeric(rafi['Bathrooms'], errors='coerce')

# Menggunakan lambda function untuk mengubah nilai kolom 'Bathrooms' menjadi 'large' atau 'small'
rafi['Bathrooms'] = rafi['Bathrooms'].apply(lambda x: 'large' if x > 2 else 'small')

# Menampilkan DataFrame setelah modifikasi
print(rafi)
```

```python
import numpy as np
# Menambahkan kolom 'newvariable' berdasarkan kondisi kolom 'Offers'
rafi['newvariable'] = np.where(rafi['Offers'] > 2, 'large', 'small')

# Menampilkan DataFrame 'rafi' setelah penambahan kolom baru
print(rafi)
```

```python
# Menambahkan kolom baru 'newvariable' yang merupakan hasil pembagian Price dengan SqFt
rafi['newvariable'] = rafi['Price'] / rafi['SqFt']

# Menampilkan DataFrame 'rafi' setelah penambahan kolom baru
print(rafi)
```

```python
# Menghapus kolom 'newvariable' dari DataFrame
rafi = rafi.drop(columns=['newvariable'])

# Menampilkan DataFrame 'rafi' setelah kolom dihapus
print(rafi)
```

```python
# Mengambil kolom 1 dan 2 dari DataFrame
kolom1_dan2 = rafi.iloc[:, 0:2]

# Menampilkan DataFrame kolom1_dan2
print(kolom1_dan2)
```

```python
# Mengganti nilai 'large' dengan 3 dan 'small' dengan 2
rafi['Bathrooms'] = rafi['Bathrooms'].replace({'large': 3, 'small': 2})

# Menampilkan DataFrame setelah modifikasi
print(rafi)
```

```python
# Memilih kolom 3 dan 4 dari DataFrame rafi
kolom3dan4 = rafi.iloc[:, 2:4]

# Menampilkan DataFrame kolom3dan4
print(kolom3dan4)
```

```python
# Menggabungkan dua DataFrame kolom1dan2 dan kolom3dan4
kolom1sd4 = pd.concat([kolom1_dan2, kolom3dan4], axis=1)

# Menampilkan DataFrame kolom1sd4
print(kolom1sd4)
```

```python
# Mengambil baris 1 sampai 3 dari DataFrame rafi
baris1sd3 = rafi.iloc[0:3, :]

# Mengambil baris 4 sampai 6 dari DataFrame rafi
baris4sd6 = rafi.iloc[3:6, :]

# Menggabungkan baris1sd3 dan baris4sd6 menggunakan pd.concat
baris1sd6 = pd.concat([baris1sd3, baris4sd6])

# Menampilkan DataFrame baris1sd6
print(baris1sd6)
```

```python
# Mengurutkan DataFrame berdasarkan kolom 'Price'
data_rafi_sort = rafi.sort_values(by='Price')

# Menampilkan DataFrame yang telah diurutkan
print(data_rafi_sort)
```
